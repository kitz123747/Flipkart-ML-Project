## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
# Check missing values
dataset.isnull().sum()

# Numeric columns → Median Imputation
num_cols = ['connected_handling_time', 'Item_price']
for col in num_cols:
    dataset[col].fillna(dataset[col].median(), inplace=True)

# Categorical columns → Mode Imputation
cat_cols = ['channel_name', 'category', 'Sub-category', 'Customer_City',
            'Product_category', 'Agent_name', 'Supervisor', 'Manager',
            'Tenure Bucket', 'Agent Shift']

for col in cat_cols:
    dataset[col].fillna(dataset[col].mode()[0], inplace=True)

# Verify again
dataset.isnull().sum()
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

I used the following techniques:

Mean/Median imputation for numeric columns → Because handling_time and item_price had skewness, median was appropriate.

Mode imputation for categorical columns → Because categorical features like channel_name, category cannot use mean/median.

Dropping rows only if missing values were <5% → To avoid unnecessary loss of information

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
 #Detect & Treat Outliers using IQR (Winsorization)

num_cols = ['connected_handling_time', 'Item_price']

for col in num_cols:
    Q1 = dataset[col].quantile(0.25)
    Q3 = dataset[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Capping
    dataset[col] = np.where(dataset[col] < lower_bound, lower_bound, dataset[col])
    dataset[col] = np.where(dataset[col] > upper_bound, upper_bound, dataset[col])
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

IQR method applied to :

connected_handling_time

item_price

Because these numerical variables had extreme values that could distort model training.

Outliers were capped (winsorization) instead of removed → To preserve dataset size.

### ML Model - 1

````python
# ML Model - 1 Implementation

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize model
lr = LogisticRegression(max_iter=500)

# Fit the algorithm
lr.fit(X_train, y_train)

# Predict on test data
y_pred_lr = lr.predict(X_test)

# Evaluation metrics
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_lr))
     
````

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.01, 0.1, 1, 10],
    'solver': ['liblinear', 'lbfgs']
}

grid_lr = GridSearchCV(LogisticRegression(max_iter=500), param_grid, cv=5, scoring='accuracy')
grid_lr.fit(X_train, y_train)

best_lr = grid_lr.best_estimator_

print("Best Parameters:", grid_lr.best_params_)
print("Best Score:", grid_lr.best_score_)

# Predict with tuned model
y_pred_lr_tuned = best_lr.predict(X_test)

print("\nTuned Model Accuracy:", accuracy_score(y_test, y_pred_lr_tuned))
     
````

#### 3. Explain each evaluation metric's indication towards business and the business impact pf the ML model used.

Accuracy: Shows overall correctness of the model. Higher accuracy means fewer incorrect predictions, leading to better customer issue handling.

Precision: Shows how many predicted “satisfied/unsatisfied” cases were actually correct. High precision means fewer false positives → avoids unnecessary escalations and saves operational cost.

Recall: Shows how many actual dissatisfied customers the model correctly identified. High recall is critical because missing unhappy customers can hurt customer retention and brand trust.

F1-Score: Balances precision and recall. High F1 means the model is reliable and stable in identifying dissatisfaction without too many wrong alerts.

Business Impact: A model with strong recall & F1 helps Flipkart detect at-risk customers early, improve agent intervention, reduce complaint escalation, and enhance overall customer experience.
